```python
import pandas as pd
```

Read files

```python
df0 = pd.read_parquet("doha_parsed_cases/all_cases.parquet")
df1 = pd.read_parquet("doha_parsed_cases/all_cases_1.parquet")
df2 = pd.read_parquet("doha_parsed_cases/all_cases_2.parquet")
```

DF Shapes

```python
print("Shape of dataset 0 is: " + str(df0.shape))
print("Shape of dataset 1 is: " + str(df1.shape))
print("Shape of dataset 2 is: " + str(df2.shape))
```

**Dataset 0**

```python

print("Columns in dataset 0 are: " + str(df0.columns.tolist()))
print("Dataset 0 first 10 rows:" + str(df0.head(10)))
```

**Dataset 1**

```python

print("Columns in dataset 1 are: " + str(df1.columns.tolist()))
print("Dataset 1 first 10 rows:" + str(df1.head(10)))
print("Dataset 1 types are: " + str(df1.dtypes))
```

```python
print("Dataset 1 missing values are: " + str(df1.isna().sum()))
print("Duplicate rows in dataset 1 are: " + str(df1["case_number"].duplicated().sum()))
```

```python
print("Dataset 1 outcome value counts are: " + str(df1["outcome"].value_counts()))
print("Dataset 1 outcome value counts normalized are: " + str(df1["outcome"].value_counts(normalize=True)))
```

```python
print("Dataset 1 unique values in 'outcome' column are: " + str(df1["outcome"].unique()))
print("Dataset 1 unique values in 'case_number' column are: " + str(df1["case_number"].unique()))
```

```python
print("Dataset 1 full text length is: " + df1["full_text"].str.len().sum().astype(str))
print("Dataset 1 full text length description is: " + str(df1["full_text"].str.len().describe()))
```

```python
print("Dataset 1 date range is: " + str(df1["date"].min()) + " to " + str(df1["date"].max()))
print("Dataset 1 date range description is: " + str(df1["date"].describe()))
print("Dataset 1 date column is: " + df1["date"].head(10).astype(str))
```

```python
print("Dataset 1 guidelines are: " + df1["guidelines"].head(10).astype(str))
print("Dataset 1 mitigating factors are: " + df1.loc[
    df1["mitigating_factors"].fillna("").str.len().nlargest(10).index,
    ["case_number", "mitigating_factors"]
])
```

**Dataset 2**

```python

print("Columns in dataset 2 are: " + str(df2.columns.tolist()))
print("Dataset 2 first 10 rows:" + str(df2.head(10)))
print("Dataset 2 types are: " + str(df2.dtypes))
```

```python
print("Dataset 2 missing values are: " + str(df2.isna().sum()))
print("Duplicate rows in dataset 2 are: " + str(df2["case_number"].duplicated().sum()))
```

```python
print("Dataset 2 outcome value counts are: " + str(df2["outcome"].value_counts()))
print("Dataset 2 outcome value counts normalized are: " + str(df2["outcome"].value_counts(normalize=True)))
```

```python
print("Dataset 2 unique values in 'outcome' column are: " + str(df2["outcome"].unique()))
print("Dataset 2 unique values in 'case_number' column are: " + str(df2["case_number"].unique()))
```

```python
print("Dataset 2 full text length is: " + df2["full_text"].str.len().sum().astype(str))
print("Dataset 2 full text length description is: " + str(df2["full_text"].str.len().describe()))
```

```python
print("Dataset 2 date range is: " + str(df2["date"].min()) + " to " + str(df2["date"].max()))
print("Dataset 2 date range description is: " + str(df2["date"].describe()))
print("Dataset 2 date column is: " + df2["date"].head(10).astype(str))
```

```python
print("Dataset 2 guidelines are: " + df2["guidelines"].head(10).astype(str))
print("Dataset 2 mitigating factors are: " + df2.loc[
    df2["mitigating_factors"].fillna("").str.len().nlargest(10).index,
    ["case_number", "mitigating_factors"]
])
   
```
